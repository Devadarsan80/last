**Problem Statement**

In the banking domain, predicting customer behavior is essential for improving business strategies and personalizing services. This project aims to build a machine learning model to predict whether a customer will perform a specific transaction in the future. The dataset contains 200 anonymized features along with a binary target variable, where 0 indicates the customer will not make a transaction and 1 indicates they will. The objective is to analyze the dataset, build and compare multiple predictive models, evaluate them using metrics suitable for imbalanced data such as F1-score and ROC-AUC, and identify the best model for deployment. This will enable the bank to proactively target potential customers, enhance engagement, and optimize business outcomes.

```python
x=data.drop('target',axis=1)
y=data['target']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y, random_state=42)
```

Observation:

Stratified splitting preserved the original class distribution in both training and test datasets. This ensures reliable and unbiased model evaluation.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score


x_train_numeric = x_train.drop('ID_code', axis=1)
x_test_numeric = x_test.drop('ID_code', axis=1)

lr = LogisticRegression(max_iter=1000)
lr.fit(x_train_numeric, y_train)

lr_pred = lr.predict(x_test_numeric)
lr_prob = lr.predict_proba(x_test_numeric)[:,1]

print("Logistic Regression")
print(classification_report(y_test, lr_pred))
print("ROC-AUC:", roc_auc_score(y_test, lr_prob))
```

Observation:

Logistic Regression achieved a strong ROC-AUC score, indicating good probability ranking. However, its recall for the minority class was limited, making it insufficient for the business objective.

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
x_train_res, y_train_res = smote.fit_resample(x_train_numeric, y_train)
```

Observation:

SMOTE successfully balanced the training dataset without introducing data leakage. This improved the model’s ability to learn patterns from minority-class samples.

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=300,
    max_depth=15,
    min_samples_leaf=5,
    class_weight='balanced',
    random_state=42)
model.fit(x_train_res, y_train_res)
model_pred = model.predict(x_test_numeric)
model_prob = model.predict_proba(x_test_numeric)[:,1]
model_pred_05 = (model_prob >= 0.5).astype(int)


print("Random Forest (Threshold = 0.5)")
print(classification_report(y_test, model_pred_05))
print("ROC-AUC:", roc_auc_score(y_test, model_prob))
```

Observation:

The Random Forest model captured non-linear relationships among features and produced reliable probability outputs. Training on balanced data improved sensitivity toward transaction-prone customers.

Observation:

Using the default threshold resulted in very low recall for the minority class. Most potential transacting customers were missed, limiting business usefulness.

```python
model_pred_015 = (model_prob >= 0.15).astype(int)

print("Random Forest (Threshold = 0.15)")
print(classification_report(y_test, model_pred_015))
```

Observation:

Lowering the threshold significantly increased recall to 100%, ensuring no potential customer was missed. This came at the cost of precision, which is acceptable in marketing use cases.

```python
from sklearn.metrics import (
    classification_report,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score
)
comparison = pd.DataFrame({
    "Threshold": ["0.5", "0.15"],
    "Precision": [
        precision_score(y_test, model_pred_05),
        precision_score(y_test, model_pred_015)
    ],
    "Recall": [
        recall_score(y_test, model_pred_05),
        recall_score(y_test, model_pred_015)
    ],
    "F1 Score": [
        f1_score(y_test, model_pred_05),
        f1_score(y_test, model_pred_015)
    ]
})

comparison
```

Observation:

The comparison clearly demonstrates the precision–recall trade-off. Threshold optimization led to improved F1-score and better alignment with business goals.

```python
model_summary = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "ROC-AUC": [
        roc_auc_score(y_test, lr_prob),
        roc_auc_score(y_test, model_prob)
    ]
})

model_summary
```

Observation:

Despite a lower ROC-AUC, the Random Forest model with threshold optimization delivered superior minority-class recall. This model was therefore selected for production deployment.

Challenges Faced

The major challenges were the absence of interpretable feature names and severe class imbalance in the target variable. These were addressed by relying on statistical validation instead of visual EDA, applying SMOTE only on the training data, and optimizing the prediction threshold to improve minority-class recall.

Conclusion

A Random Forest model trained on SMOTE-balanced data and deployed with a custom probability threshold of 0.15 was selected as the final solution. Although Logistic Regression achieved a higher ROC-AUC score, the chosen model delivered 100% recall for potential transacting customers, making it more suitable for real-world banking and marketing applications.
